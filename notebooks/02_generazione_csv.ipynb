```python
import os
import pandas as pd

raw_root = paths['raw_images']

rows = []
for esperimento in sorted(os.listdir(raw_root)):
    esp_path = os.path.join(raw_root, esperimento)
    if not os.path.isdir(esp_path):
        continue
    for gruppo in sorted(os.listdir(esp_path)):
        gruppo_path = os.path.join(esp_path, gruppo)
        if not os.path.isdir(gruppo_path):
            continue
        for fname in sorted(os.listdir(gruppo_path)):
            if fname.lower().endswith(('.tif', '.tiff', '.png', '.jpg')):
                rows.append({
                    'filename': fname,
                    'filepath': os.path.join(esperimento, gruppo, fname),
                    'esperimento': esperimento,
                    'gruppo_ingrandimento': gruppo,
                    'nm_per_pixel': None,
                })

df = pd.DataFrame(rows)   # <-- aggiunta: costruisce df dalla lista rows

n_combinazioni = df.groupby(['esperimento', 'gruppo_ingrandimento']).ngroups
log(f'Trovate {len(df)} immagini in {n_combinazioni} combinazioni esperimento+gruppo distinte')

riepilogo = df.groupby(['esperimento', 'gruppo_ingrandimento']).size().reset_index(name='n_immagini')
print(riepilogo)
df.head()
```

```python
import numpy as np

tolleranza_relativa = 0.03  # 3% di differenza massima per considerare due misure "la stessa calibrazione"

valori_unici = sorted(set(misure.values()))

clusters = []
current_cluster = [valori_unici[0]]
for v in valori_unici[1:]:
    if (v - current_cluster[-1]) / current_cluster[-1] <= tolleranza_relativa:
        current_cluster.append(v)
    else:
        clusters.append(current_cluster)
        current_cluster = [v]
clusters.append(current_cluster)

cluster_map = {}
for i, cluster in enumerate(clusters):
    media = np.mean(cluster)
    label = f'classe_{i+1}'
    for v in cluster:
        cluster_map[v] = (label, media)

print(f"Da {len(misure)} combinazioni esperimento-gruppo, individuate {len(clusters)} classi di calibrazione reali:")
for i, cluster in enumerate(clusters):
    print(f"  classe_{i+1}: valori originali {cluster} -> media {np.mean(cluster):.3f} nm/px")
```

```python
misure_classe = {k: cluster_map[v][0] for k, v in misure.items()}
misure_valore_finale = {k: cluster_map[v][1] for k, v in misure.items()}

df['classe_calibrazione'] = df.apply(
    lambda r: misure_classe.get((r['esperimento'], r['gruppo_ingrandimento'])), axis=1
)
df['nm_per_pixel'] = df.apply(
    lambda r: misure_valore_finale.get((r['esperimento'], r['gruppo_ingrandimento'])), axis=1
)

csv_path = os.path.join(paths['annotations'], 'calibrazione.csv')
df.to_csv(csv_path, index=False)
log('calibrazione.csv aggiornato con classi di calibrazione reali (indipendenti dall\'esperimento)')

df.groupby('classe_calibrazione')['nm_per_pixel'].agg(['mean', 'count'])
```

```python
riepilogo_classi = df.groupby('classe_calibrazione').agg(
    n_immagini=('filename', 'count'),
    nm_per_pixel=('nm_per_pixel', 'first'),
    esperimenti_coinvolti=('esperimento', lambda x: sorted(x.unique())),
    gruppi_originali=('gruppo_ingrandimento', lambda x: sorted(x.unique()))
).sort_values('nm_per_pixel')

print(f"Totale classi di calibrazione reali: {df['classe_calibrazione'].nunique()}")
print(f"Totale immagini: {len(df)}\n")

for classe, row in riepilogo_classi.iterrows():
    print(f"{classe}: {row['nm_per_pixel']:.3f} nm/px, {row['n_immagini']} immagini")
    print(f"   esperimenti: {row['esperimenti_coinvolti']}")
    print(f"   etichette originali: {row['gruppi_originali']}\n")
```
