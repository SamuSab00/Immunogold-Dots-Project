**PyTorch** --> PyTorch è un framework di deep learning open source basato su Python e C++, usato per creare e addestrare reti neurali. Un framework (letteralmente "struttura di lavoro") è un insieme di codice predefinito, librerie e strumenti che fornisce la base strutturale per sviluppare applicazioni software.

**Scikit-Image** --> libreria open-source per il linguaggio di programmazione Python dedicata all'elaborazione delle immagini. Viene utilizzata per analizzare, manipolare ed estrarre dati da immagini digitali in contesti di ricerca, industria e intelligenza artificiale.

**opencv-python-headless** -->La versione di OpenCV (Open Source Computer Vision Library) priva di moduli per l'interfaccia grafica (GUI). Ideale per ambienti senza schermo come server, container Docker o notebook (es. Google Colab), evitando errori dovuti alla mancanza di dipendenze di sistema visive (X11/Qt).

````python
import datetime

log_path = os.path.join(base, 'logs', 'project_log.txt')

def log(msg):
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M')
    with open(log_path, 'a') as f:
        f.write(f'[{timestamp}] {msg}\n')
    print(f'[{timestamp}] {msg}')

log('Ambiente Colab inizializzato, struttura cartelle creata')
````

Da qui in poi, ogni volta che completi un passaggio importante (es. "annotazione completata per 20 immagini", "primo training avviato"), chiami log('...') e hai una cronologia scritta in logs/project_log.txt, persistente su Drive.

````python
%%writefile /content/drive/MyDrive/progetto_immunogold/scripts/common.py
import os
import datetime
import torch

base = '/content/drive/MyDrive/progetto_immunogold'

paths = {
    'raw_images': os.path.join(base, 'raw_images'),
    'annotations': os.path.join(base, 'annotations'),
    'patches_images': os.path.join(base, 'patches', 'images'),
    'patches_density': os.path.join(base, 'patches', 'density_maps'),
    'checkpoints': os.path.join(base, 'checkpoints'),
    'logs': os.path.join(base, 'logs'),
}

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def log(msg):
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M')
    log_path = os.path.join(paths['logs'], 'project_log.txt')
    with open(log_path, 'a') as f:
        f.write(f'[{timestamp}] {msg}\n')
    print(f'[{timestamp}] {msg}')
````
